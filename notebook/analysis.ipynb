### Pizza categories generate the most revenue

```python
df_pizza_orders = pd.merge(df_pizza, df_order_details, on="pizza_id", how="inner")

df_pizza_orders.head()
```

```python
df_pizza_orders.isnull().sum()
```

```python
df_pizza_orders.isna().sum()
```

```python
df_pizza_orders["Revenue"] =  df_pizza_orders["price"] * df_pizza_orders["quantity"]

df_pizza_orders.head()
```

```python
Total_Revenue = df_pizza_orders["Revenue"].sum()
round(Total_Revenue, 2)
```

### Revenue by pizza category

```python
df_pizza_category = pd.merge(df_pizza_types, df_pizza_orders, on="pizza_type_id", how="outer")

df_pizza_category.head()
```

```python
df_pizza_category.isnull().sum()
```

```python
df_pizza_category.isna().sum()
```

```python
Category_Revenue = df_pizza_category.groupby("category")["Revenue"].sum()
Category_Revenue
```

### Analyze sales volume

```python
Sales_Volumes = df_pizza_category.groupby("category")["quantity"].sum()
Sales_Volumes
```

### Average Selling Price

```python
Average_Selling_Price = df_pizza["price"].mean()
Average_Selling_Price 
```

### Average selling price by category

```python
Average_Selling_Price_Category = (
    df_pizza_category.groupby("category")["Revenue"].sum()
    / df_pizza_category.groupby("category")["quantity"].sum()
)

Average_Selling_Price_Category
```

### Top 5 pizza types by revenue

```python
Top_5_Pizza = df_pizza_category.groupby("name")["Revenue"].sum().sort_values(ascending=False).head(5)
Top_5_Pizza
```

### Find the bottom 5 pizza types by revenue

```python
Bottom_5_Pizza = df_pizza_category.groupby("name")["Revenue"].sum().sort_values().head(5)
Bottom_5_Pizza
```

### Top 5 pizza types by quantity sold

```python
Top_5_Pizza_Quantity = df_pizza_category.groupby("name")["quantity"].sum().sort_values(ascending=False).head(5)
Top_5_Pizza_Quantity
```

### Revenue by Month

```python
df_order_pizza = pd.merge(df_orders, df_pizza_category, on="order_id", how="outer")

df_order_pizza.head()
```

```python
df_order_pizza["Month"] = df_order_pizza["date"].dt.to_period("M")

df_order_pizza.head(5)
```

```python
Monthly_Revenue = df_order_pizza.groupby("Month")["Revenue"].sum()
Monthly_Revenue
```

### Monthly revenue growth

```python
Monthly_Changed_Revenue = Monthly_Revenue.diff()
Monthly_Changed_Revenue
```

```python
Monthly_Growth_Revenue = Monthly_Revenue.pct_change() * 100
Monthly_Growth_Revenue
```

### Busiest month by order count

```python
Order_Count_BYMonth =  df_order_pizza.groupby("Month")["order_id"].nunique()
Order_Count_BYMonth
```

### Popular pizza Size by category

```python
Popular_Size_byCategory = df_order_pizza.groupby(["category", "size"])["quantity"].sum()
Popular_Size_byCategory
```

### AOV by Month

```python
AOV_Month = Monthly_Revenue / Order_Count_BYMonth
AOV_Month
```

### Pizza category contributes to total revenue

```python
Category_Contribution_Revenue = Category_Revenue / Total_Revenue * 100
Category_Contribution_Revenue
```

### Weekdays VS Weekend Sales

```python
df_order_pizza["days_of_week"] = df_order_pizza["date"].dt.dayofweek.apply(
    lambda x: "Weekend" if x >= 5 else "Weekday"
)

df_order_pizza.head()
```

```python
Quantity_By_Days_of_Week = df_order_pizza.groupby("days_of_week")["quantity"].sum()
Quantity_By_Days_of_Week 
```

### Average daily sales by day type

```python
(
    df_pizza_category.groupby("category")["Revenue"].sum()
    / df_pizza_category.groupby("category")["quantity"].sum()
)
```

```python
Average_Daily_Sales = df_order_pizza.groupby(["date", "days_of_week"])["quantity"].sum().groupby("days_of_week").mean()
Average_Daily_Sales
```

# Insights:

1. Lowest Month by Revenue is October it also has the lowest orders. so we could try some offers during this month.

2. Classic made the most revenue and sold the most pizza, while Veggie made the least revenue. Chicken had the fewest pizzas sold.

3. Even though Chicken sold the least pizza it was the highest pizza sold in every category while Veggie was the least category what sold pizza and also made the least revenue.

4. We can give combo offers on Veggies Pizza and see if customers give good response on Sales as well as on revenues.
